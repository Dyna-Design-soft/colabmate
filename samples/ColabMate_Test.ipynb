## R6 — Make the stats bars move

Watch the **RAM** (and **GPU RAM** on a GPU runtime) bars on the app's Colab screen while this runs for ~30 s.

```python
import time
import numpy as np

blocks = []
for i in range(6):
    blocks.append(np.ones((2500, 10000)))  # ~200 MB each
    print(f'allocated {(i + 1) * 200} MB RAM')
    time.sleep(3)

try:
    import torch
    if torch.cuda.is_available():
        t = torch.randn(8000, 8000, device='cuda')
        for i in range(10):
            t = t @ t / 8000          # keeps the GPU busy
            torch.cuda.synchronize()
            print(f'GPU step {i + 1}/10 — GPU memory {torch.cuda.memory_allocated() / 1e9:.1f} GB')
            time.sleep(1)
        del t
        torch.cuda.empty_cache()
    else:
        print('No GPU — only RAM will move')
except ImportError:
    print('PyTorch not installed — only RAM will move')

del blocks
print('released')
```

## Long run — fake training loop (good for watching live output)

```python
import time, random

loss = 2.5
for epoch in range(1, 21):
    loss *= random.uniform(0.85, 0.97)
    acc = min(0.99, 1 - loss / 3)
    print(f'epoch {epoch:02d}/20  loss={loss:.4f}  acc={acc:.3f}')
    time.sleep(2)
print('training finished ✅')
```
